```python
import pandas as pd
```

```python
titanic = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')
```

```python
titanic['Age']
```

```python
titanic['Age'].mean()
```

```python
titanic['Age'].std()
```

```python
s = (titanic['Age'] - titanic['Age'].mean()) / titanic['Age'].std()
```

```python
import seaborn as sns
```

```python
sns.kdeplot(s)
```

```python
s.mean()  # mean reached near to 0 
```

```python
s.std() # std dev reached to 1
```

```python
#Hence, converted to standard normal variate
```
