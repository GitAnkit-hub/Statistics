```python
import pandas as pd 
import numpy as np
```

* dice rolling bar chart

```python
import random
```

```python
l = []
for i in range(10000):
    l.append(random.randint(1,6))
```

```python
l[:5]
```

```python
s = (pd.Series(l).value_counts()/pd.Series(l).value_counts().sum()).sort_index()

s
```

```python
# s.plot(kind='bar')
s.plot()
```

* rolling dice twice

```python
l = []
for i in range(10000):
    l.append(random.randint(1,6) + random.randint(1,6))
```

```python
l[:10]
```

```python
s = (pd.Series(l).value_counts()/pd.Series(l).value_counts().sum()).sort_index()
s
```

```python
s.plot(kind='bar')

# prob of being 7 sum is : 6/36
```

* cumulative distribution function

```python
# single dice roll
n = np.cumsum(s)
```

```python
n.plot()
```

```python
n.plot(kind='bar')
```
