```python
import seaborn as sns
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
sns.kdeplot(data=df, x='sepal_length', hue='species')
```

```python
sns.kdeplot(data=df, x='sepal_width', hue='species')
```

```python
sns.kdeplot(data=df, x='petal_length', hue='species')
```

```python
sns.kdeplot(data=df, x='petal_width', hue='species')
```

* Here, we can see that petal_length and petal_width are the features that can distinguish the 3 categories easily
    * if petal_length <= 2.3 -> this is setosa
    * 2.3<= petal_length <=5 -> there is a higher probability density of being versicolor.
    * and from 5 onwards, it is virginica higher prob density.

* we are able to distinguish the catergories based on a single feature itself "__petal_length__"
* and PDF helped us in making this conclusion.

* like wise, __petal_width__ is also one other feature.

* more overlap in  __sepal_width__ , it is not a good distinguishing feature.
* less overlap in __sepal_length__ than __sepal_width__ but still can disitnguish.

- lets see CDF also

```python
sns.kdeplot(data=df, x="petal_length", hue="species")
sns.ecdfplot(data=df, x="petal_length", hue="species")
```

* here, this tells us the probability density for a point. 
* when petal_length =4.9, only nearly 10 % of the viriginica flowers follow this but while 100% of the versicolor flowers follow this.

* take petal_length=4.5, merely viriginica flowers have the this petal_length while this petal_length is there in 80% of the versicolor flowers.
