```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from numpy.random import normal
```

```python
sample = normal(loc=50, scale=5, size=1000)

# mean = 50
#std dev = 5
```

```python
sample
```

```python
sample.mean()

# this will not be exactly 5o since , we are haveing sample data points not complete data points.
```

```python
#plot histogram to understand the distribution of data

plt.hist(sample,bins=10)
```

```python
# looking at histrogram , we concluded that the distribution looks like normal distribution.
# so, we will calculate the parameters like mean, std dev.
# fit the distribution with the parametrs we get to the known ditribution pdf function

# then i will give data points to the pdf and it will return the density of each point.
```

```python
#calculate sample mean and sample std dev

sample_mean = sample.mean()
sample_std = sample.std()
```

```python
#fit the distribution with the above parametrs

from scipy.stats import norm
dist = norm(sample_mean, sample_std)
```

```python
values = np.linspace(sample.min(), sample.max(),100)
```

```python
# prob density i am getting
# using pdf function
prob_density = [dist.pdf(value) for value in values]
```

```python
#plot the histrogram  and pdf

plt.hist(sample,bins=10, density=True)
plt.plot(values, prob_density)
```
